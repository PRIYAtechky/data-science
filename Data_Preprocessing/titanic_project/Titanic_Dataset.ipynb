```python
import matplotlib.pyplot as plt

df["Survived"].value_counts().plot(kind="bar")

plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.title("Survival Count")
plt.show()
```

```python
pd.crosstab(df["Sex"], df["Survived"])
```

```python
pd.crosstab(df["Sex"], df["Survived"]).plot(kind="bar")

plt.xlabel("Sex")
plt.ylabel("Number of Passengers")
plt.title("Survival by Gender")
plt.show()
```

```python
pd.crosstab(df["Pclass"], df["Survived"])
```

```python
pd.crosstab(df["Pclass"], df["Survived"]).plot(kind="bar")

plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.title("Survival by Passenger Class")
plt.show()
```

```python
df["Age"].hist()

plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title("Age Distribution")
plt.show()
```

```python
df["Fare"].hist()

plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.title("Fare Distribution")
plt.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
# 1. Survival Count
fig1 = px.bar(
    df,
    x="Survived",
    title="Survival Count"
)

# 2. Survival by Gender
fig2 = px.histogram(
    df,
    x="Sex",
    color="Survived",
    barmode="group",
    title="Survival by Gender"
)

# 3. Survival by Passenger Class
fig3 = px.histogram(
    df,
    x="Pclass",
    color="Survived",
    barmode="group",
    title="Survival by Passenger Class"
)

# 4. Age Distribution
fig4 = px.histogram(
    df,
    x="Age",
    title="Age Distribution"
)

# 5. Fare Distribution
fig5 = px.histogram(
    df,
    x="Fare",
    title="Fare Distribution"
)

# 6. Age vs Fare
fig6 = px.scatter(
    df,
    x="Age",
    y="Fare",
    color="Survived",
    title="Age vs Fare"
)
```

```python
from plotly.subplots import make_subplots

dashboard = make_subplots(
    rows=3,
    cols=2,
    subplot_titles=(
        "Survival Count",
        "Survival by Gender",
        "Survival by Passenger Class",
        "Age Distribution",
        "Fare Distribution",
        "Age vs Fare"
    )
)

dashboard.add_trace(fig1.data[0], row=1, col=1)

for trace in fig2.data:
    dashboard.add_trace(trace, row=1, col=2)

for trace in fig3.data:
    dashboard.add_trace(trace, row=2, col=1)

dashboard.add_trace(fig4.data[0], row=2, col=2)

dashboard.add_trace(fig5.data[0], row=3, col=1)

for trace in fig6.data:
    dashboard.add_trace(trace, row=3, col=2)

dashboard.update_layout(
    height=1000,
    width=1000,
    title_text="Titanic Data Analysis Dashboard",
    showlegend=True
)

dashboard.show()
```
