#Data Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))

sns.lineplot(data=df, x='Experience', y='Salary', marker='o')

plt.title("Experience vs Salary - Line Plot")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

```python
sns.histplot(data=df, x='Salary', kde=True)
plt.title("Salary Distribution")
plt.show()
```

#Feature Scaling

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
standard_scaler = StandardScaler()

standard_scaled = standard_scaler.fit_transform(
    df[['Experience', 'Salary']]
)

print(type(standard_scaled))

standard_df = pd.DataFrame(
    standard_scaled,
    columns=['Experience', 'Salary']
)

print("Standard Scaled Data")
print(standard_df.head())
```

```python
minmax_scaler = MinMaxScaler()

minmax_scaled = minmax_scaler.fit_transform(
    df[['Experience', 'Salary']]
)

minmax_df = pd.DataFrame(
    minmax_scaled,
    columns=['Experience', 'Salary']
)

print("MinMax Scaled Data")
print(minmax_df.head())
```
